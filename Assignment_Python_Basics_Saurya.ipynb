7. What is the difference between “==” and “is” operators in Python
-> "==" checks if values are equal.

  "is"  checks if two variables point to the same object in memory.

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True (values same)
print(a is b)  # False (different memory locations)
```

10. What is the difference between implicit and explicit type casting
->Python automatically converts one data type to another without you telling it to when it needs to prevent data loss.

Why:
This usually happens in mathematical operations where two different data types are used — Python converts the smaller/less precise type to the bigger/more precise one to avoid losing information.

You manually tell Python to convert a value from one data type to another using built-in functions like:

int() → convert to integer

float() → convert to float

str() → convert to string

bool() → convert to boolean

Why:
You need to do this when Python won’t do it automatically, or when you want to control the conversion.

```python
# Implicit
a = 5      # int
b = 2.5    # float
c = a + b  # 7.5 (float, automatic conversion)
print(c)

# Explicit
num = int("10")
print(num + 5)  # 15
```

13. What is the difference between for and while loops
-> for loop: Used when you know the number of iterations.

while loop: Used when the number of iterations is not known in advance.

```python
# for loop
for i in range(3):
    print(i)

# while loop
n = 3
while n > 0:
    print(n)
    n -= 1
```

14. Describe a scenario where a while loop is more suitable than a for loop.
-> Consider a program that asks a user to enter the correct password. Since we do not know how many attempts the user will take, it is not possible to set a fixed number of iterations in advance.
The while loop will keep prompting the user until the entered password matches the stored one, and only then will it stop.
This makes while loops ideal for situations involving repeated execution based on an uncertain condition, such as user authentication, waiting for a file to download, or checking for incoming data in real-time programs.

9. Write a Python program to convert user input from string to integer, float, and boolean types.

```python
# Taking user input as string
user_input = input("Enter something: ")

# Converting to integer
try:
    int_value = int(user_input)
    print("Integer:", int_value)
except ValueError:
    print("Cannot convert to integer")

# Converting to float
try:
    float_value = float(user_input)
    print("Float:", float_value)
except ValueError:
    print("Cannot convert to float")

# Converting to boolean
bool_value = bool(user_input)  # Empty string becomes False, otherwise True
print("Boolean:", bool_value)
```

14. Write a program to reverse a string using a while loop.

```python
# Taking user input
text = input("Enter a string: ")

# Initializing variables
reversed_text = ""
index = len(text) - 1

# Using while loop to reverse
while index >= 0:
    reversed_text += text[index]
    index -= 1

print("Reversed string:", reversed_text)
```

15. Write a Python program to calculate the factorial of a number provided by the user using a while loop

```python
# Taking user input
num = int(input("Enter a number: "))

factorial = 1
i = 1

# Calculating factorial using while loop
while i <= num:
    factorial *= i
    i += 1

print(f"Factorial of {num} is:", factorial)
```
